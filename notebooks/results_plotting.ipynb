```python
def average_every_n(input_list, n):
    result = []
    for i in range(0, len(input_list), n):
        chunk = input_list[i:i+n]
        if chunk:
            avg = sum(chunk) / len(chunk)
            result.append(avg)
    return result
```

```python
def plot_ft(ft_json, orig_acc):
    fig, axes = plt.subplots(nrows = 1, ncols = 3, figsize = (15, 4), dpi = 100)
    
    train_losses = np.array(ft_json['train_losses'])
    x = np.array([i for i in range(len(train_losses))])
    axes[0].plot(x, train_losses)
    axes[0].set_title('Training Loss Over Steps')
    axes[0].set_xlabel('Training Steps')
    axes[0].set_ylabel('Training Loss')

    val_losses = np.array(ft_json['val_losses'][1:])
    epochs = np.array([i for i in range(len(val_losses))])
    axes[1].plot(epochs, val_losses)
    axes[1].set_title('Validation Loss Over Epochs')
    axes[1].set_xlabel('Epochs')
    axes[1].set_ylabel('Validation Loss')

    accs = np.array(ft_json['top1_accs'])
    epochs = np.array([i for i in range(len(accs))])
    axes[2].plot(epochs, accs)
    axes[2].set_title('Top-1 Accuracy Over Epochs')
    axes[2].set_xlabel('Epochs')
    axes[2].set_ylabel('Top-1 Accuracy')
    axes[2].axhline(orig_acc, linestyle = '--', color = 'red')
    plt.show()
```

```python
def compare_plot_ft(ft_json1, ft_json2, orig_acc, label1, label2):
    fig, axes = plt.subplots(nrows = 1, ncols = 3, figsize = (15, 4), dpi = 100)
    
    train_losses1 = np.array(ft_json1['train_losses'])
    train_losses2 = np.array(ft_json2['train_losses'])
    train_losses1 = train_losses1[:len(train_losses2)]
    x = np.array([i for i in range(len(train_losses2))])
    axes[0].plot(x, train_losses1, label = label1, color = 'tab:blue', alpha = 0.6)
    axes[0].plot(x, train_losses2, label = label2, color = 'tab:green', alpha = 0.6)
    axes[0].set_title('Training Loss Over Steps')
    axes[0].set_xlabel('Training Steps')
    axes[0].set_ylabel('Training Loss')
    axes[0].legend()

    val_losses1 = np.array(ft_json1['val_losses'][1:])
    val_losses2 = np.array(ft_json2['val_losses'][1:])
    val_losses1 = val_losses1[:len(val_losses2)]
    epochs = np.array([i for i in range(len(val_losses1))])
    axes[1].plot(epochs, val_losses1, label = label1, color = 'tab:blue', alpha = 0.8)
    axes[1].plot(epochs, val_losses2, label = label2, color = 'tab:green', alpha = 0.8)
    axes[1].set_title('Validation Loss Over Epochs')
    axes[1].set_xlabel('Epochs')
    axes[1].set_ylabel('Validation Loss')

    accs1 = np.array(ft_json1['top1_accs'])
    accs2 = np.array(ft_json2['top1_accs'])
    accs1 = accs1[:len(accs2)]
    epochs = np.array([i for i in range(len(accs1))])
    axes[2].plot(epochs, accs1, label = label1, color = 'tab:blue', alpha = 0.8)
    axes[2].plot(epochs, accs2, label = label2, color = 'tab:green', alpha = 0.8)
    axes[2].set_title('Top-1 Accuracy Over Epochs')
    axes[2].set_xlabel('Epochs')
    axes[2].set_ylabel('Top-1 Accuracy')
    axes[2].axhline(orig_acc, linestyle = '--', color = 'red')
    plt.show()
```

```python
from matplotlib.ticker import StrMethodFormatter
def plot_rep_sim(rep_sim_json, func = 'CKA'):
    num_cols = len(rep_sim_json)
    fig, axes = plt.subplots(nrows = 2, ncols = 7, figsize = (25, 9.5), dpi = 100)
    cmap = plt.get_cmap('viridis')
    row_idx = 0
    col_idx = 0
    colors = cmap(np.linspace(0, 1, len(rep_sim_json)))
    for i,key in enumerate(rep_sim_json):
        rep_sims = np.array(rep_sim_json[key])
        num_steps = np.array([j for j in range(len(rep_sims))])
        color = colors[i]
        axes[row_idx, col_idx].yaxis.set_major_formatter(StrMethodFormatter('{x:.1f}'))
        axes[row_idx, col_idx].plot(num_steps, rep_sims, color = color)
        if i == 0:
            axes[row_idx, col_idx].set_ylabel('1-Rep Sim')
            axes[row_idx, col_idx].set_xlabel('Training Steps')
        axes[row_idx, col_idx].set_title(f'Layer {i+1} {func}, Rank 1024')
        if col_idx == 6:
            col_idx = 0
            row_idx += 1
        else:
            col_idx += 1
    # fig.delaxes(axes[1, 3])
```

```python
def comp_plot_rep_sim(rep_sim_jsons, func = 'CKA'):
    num_cols = len(rep_sim_jsons[0])
    fig, axes = plt.subplots(nrows = 2, ncols = 4, figsize = (11, 9.5), dpi = 100)
    cmap = plt.get_cmap('viridis')
    colors = cmap(np.linspace(0, 1, num_cols))
    # cmap2 = plt.get_cmap('cividis')
    # colors2 = cmap1(np.linspace(0, 1, num_cols))
    for j, rep_sim_json in enumerate(rep_sim_jsons):
        row_idx = 0
        col_idx = 0
        for i,key in enumerate(rep_sim_json):
            rep_sims = np.array(rep_sim_json[key])
            num_steps = np.array([j for j in range(len(rep_sims))])
            color = colors[i]
            axes[row_idx, col_idx].yaxis.set_major_formatter(StrMethodFormatter('{x:.1f}'))
            if j == 1:
                linestyle = 'dotted'
                marker = 'o'
                label = 'joint'
                alpha = 0.7
            elif j == 2:
                linestyle = 'dashdot'
            else:
                linestyle = 'solid'
                marker = 's'
                label = 'Independent'
                alpha = 1
            axes[row_idx, col_idx].plot(num_steps, rep_sims, color = color, linewidth = 1, linestyle = linestyle, alpha = alpha, label = label)
            axes[row_idx, col_idx].legend()
            if i == 0:
                axes[row_idx, col_idx].set_ylabel('1-Rep Sim')
                axes[row_idx, col_idx].set_xlabel('Training Steps')
            axes[row_idx, col_idx].set_title(f'Layer {i+1} {func}, Rank 1024')
            if col_idx == 3:
                col_idx = 0
                row_idx += 1
            else:
                col_idx += 1
    fig.delaxes(axes[1, 3])
```
